```python
X=df.drop('species',axis=1)
y=df['species']
from keras.models import Sequential
from keras.layers import Dense
from keras.utils import to_categorical
df['species'] = pd.Categorical(df.species)
df['species'] = df.species.cat.codes
# Turn response variable into one-hot response vectory = to_categorical(df.response)
y = to_categorical(df.species)
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30,stratify=y,random_state=123)
model=Sequential()
model.add(Dense(100,activation='relu',input_shape=(4,)))

model.add(Dense(3,activation='softmax'))
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
history=model.fit(X_train,y_train,epochs=45,validation_data=(X_test, y_test))
model.evaluate(X_test,y_test)
```

```python
pred = model.predict(X_test[:10])
print(pred)
```

```python
# Getting predictions from the model for the given examples.
import numpy as np
predictions = model.predict(X)

# Convert one-hot encoded y and probability predictions to single class labels
actual_labels = np.argmax(y, axis=1)
predicted_labels = np.argmax(predictions, axis=1)

# Compare with the actual charges
Scores = pd.DataFrame({'Actual': actual_labels, 'Predictions': predicted_labels})
Scores.head()
```
